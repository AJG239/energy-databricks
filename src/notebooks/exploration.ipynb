## Notebook - Exploración de datos

```python
import requests as rq
```

```python
url = "https://apidatos.ree.es/es/datos/mercados/precios-mercados-tiempo-real"

params = {
    "start_date": "2025-10-01T00:00",
    "end_date": "2025-10-01T23:59",
    "time_trunc": "hour",   
    "geo_limit": "peninsular"
}

req = rq.get(
    url,
    params= params,
    timeout= 30
) 

print(req)
```

```python
data = req.json()
data.keys()
```

```python
data['data']
```

```python
data["included"]
```

```python
included = data["included"]
pvp = included[0]
values = pvp['attributes']['values']
len(values)
```

```python
len(pvp['attributes']['values'])
```

```python
list(values)
```

```python
value = [v['value'] for v in values]
value
```

```python
dates = [d['datetime'] for d in values]
dates
```
